# 🎨 ComfyUI trên Google Colab

Cơ sở để chạy ComfyUI trên Colab. Notebook mỏng — toàn bộ logic nằm trong package [`comfycolab`](https://github.com/khuongcracker/comfy-colab), test được bằng `pytest` mà không cần Colab.

**Dùng:** bấm ▶︎ ở ô bên dưới. Mặc định chỉ dựng ComfyUI, không tải model nào, khoảng 2–4 phút là có link.

Model thì tải bằng **ComfyUI-Manager** ngay trong giao diện, hoặc dán URL vào ô `Models`.

Ảnh xuất, input và setting lưu trên Google Drive (`MyDrive/ComfyData`) nên sống qua các phiên Colab.

> Fork repo về tài khoản của bạn thì nhớ sửa biến `REPO` trong ô bên dưới.


```python
#@title # 🎨 Chạy ComfyUI { display-mode: "form" }
#@markdown Bấm ▶︎ bên trái. Lần đầu khoảng 2–4 phút (chưa tải model nào).

#@markdown ### 🧩 Bộ custom node
#@markdown `fast` = ComfyUI + Manager (đủ để chạy, tự thêm node trong giao diện) ·
#@markdown `base` = thêm controlnet/upscale/inpaint · `full` = thêm Impact/IPAdapter/video ·
#@markdown `none` = ComfyUI trần
NodeSet = "fast"  #@param ["fast", "base", "full", "none"]

#@markdown ---
#@markdown ### 📁 Dữ liệu
#@markdown Ảnh xuất, input và setting luôn nằm ở đây, không mất khi Colab ngắt.
MountDrive = True  #@param {type:"boolean"}
DataDir = "/content/drive/MyDrive/ComfyData"  #@param {type:"string"}

#@markdown **Model lưu ở đâu** — `drive` giữ qua phiên nhưng Drive free chỉ 15 GB.
#@markdown `session` dùng đĩa tạm Colab (~80 GB), nhanh hơn nhưng mất khi ngắt.
ModelStorage = "drive"  #@param ["drive", "session"]

#@markdown ---
#@markdown ### 🧠 Model (tuỳ chọn)
#@markdown **Để trống là xong** — cứ mở ComfyUI lên rồi tải model bằng Manager
#@markdown trong giao diện. Muốn tải sẵn thì dán URL HuggingFace/CivitAI,
#@markdown ngăn nhau bằng dấu phẩy. Đặt tên file bằng `<url>@=ten.safetensors`.
Models = ""  #@param {type:"string"}
#@markdown Thư mục đích mặc định là `checkpoints`. Model loại khác (vae, lora,
#@markdown text_encoders...) thì thêm vào `data/models.yaml` hoặc dùng Manager.

#@markdown ---
#@markdown ### 🌐 Kết nối
Tunnel = "cloudflare"  #@param ["cloudflare", "pinggy", "none"]
Region = "auto"  #@param ["auto", "asia", "usa", "europe", "australia", "south america"]

#@markdown ---
#@markdown ### 🔑 Token — chỉ cần khi tải model bị giới hạn
CivitaiToken = ""  #@param {type:"string"}
HFToken = ""  #@param {type:"string"}

#@markdown ---
#@markdown ### ⚙️ Nâng cao
ExtraArgs = ""  #@param {type:"string"}
Branch = "main"  #@param {type:"string"}

# Nguồn code. Fork về tài khoản khác thì đổi dòng này.
REPO = "https://github.com/khuongcracker/comfy-colab"

import os
import subprocess
import sys

SRC = "/content/comfy-colab"
if not os.path.isdir(SRC):
    subprocess.run(
        ["git", "clone", "--depth", "1", "-b", Branch, REPO, SRC], check=True
    )
else:
    subprocess.run(["git", "-C", SRC, "pull", "--ff-only"], check=False)

sys.path.insert(0, f"{SRC}/src")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "pyyaml"], check=False)

from comfycolab import Config, launch

launch(
    Config.from_notebook(
        node_set=NodeSet,
        data_dir=DataDir,
        model_storage=ModelStorage,
        models=Models,
        tunnel=Tunnel,
        tunnel_region=Region,
        extra_args=ExtraArgs,
        mount_drive=MountDrive,
        civitai_token=CivitaiToken,
        hf_token=HFToken,
    )
)
```
