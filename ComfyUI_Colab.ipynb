# 🎨 ComfyUI trên Google Colab

Notebook mỏng — toàn bộ logic nằm trong package [`comfycolab`](https://github.com/khuongcracker/comfy-colab), test được bằng `pytest` mà không cần Colab.

**Dùng:** chọn preset trong form bên dưới rồi bấm ▶︎. Lần đầu 3–8 phút tuỳ preset.

Model và ảnh xuất lưu trên Google Drive (`MyDrive/ComfyData`) nên sống qua các phiên Colab.

> Fork repo về tài khoản của bạn thì nhớ sửa biến `REPO` trong ô bên dưới.


```python
#@title # 🎨 Cài đặt & chạy ComfyUI { display-mode: "form" }
#@markdown Chọn preset rồi bấm ▶︎ bên trái. Lần đầu mất khoảng 3–8 phút tuỳ preset.

Preset = "flux"  #@param ["fast", "sd15", "sdxl", "flux", "full"]

#@markdown ---
#@markdown ### 📁 Dữ liệu
#@markdown Model và ảnh xuất nằm ở đây, không mất khi Colab ngắt.
MountDrive = True  #@param {type:"boolean"}
DataDir = "/content/drive/MyDrive/ComfyData"  #@param {type:"string"}

#@markdown ---
#@markdown ### 🧩 Model tải thêm
#@markdown Tên có trong `data/models.yaml`, hoặc dán URL HuggingFace/CivitAI.
#@markdown Nhiều thứ thì ngăn bằng dấu phẩy.
ExtraModels = ""  #@param {type:"string"}

#@markdown ---
#@markdown ### 🌐 Kết nối
Tunnel = "cloudflare"  #@param ["cloudflare", "pinggy", "none"]
Region = "auto"  #@param ["auto", "asia", "usa", "europe", "australia", "south america"]

#@markdown ---
#@markdown ### 🔑 Token
#@markdown Chỉ cần khi tải model bị giới hạn. Để trống nếu không cần.
CivitaiToken = ""  #@param {type:"string"}
HFToken = ""  #@param {type:"string"}

#@markdown ---
#@markdown ### ⚙️ Nâng cao
ExtraArgs = ""  #@param {type:"string"}
Branch = "main"  #@param {type:"string"}

# Nguồn code. Fork về tài khoản khác thì đổi dòng này.
REPO = "https://github.com/khuongcracker/comfy-colab"

import os
import subprocess
import sys

SRC = "/content/comfy-colab"
if not os.path.isdir(SRC):
    subprocess.run(
        ["git", "clone", "--depth", "1", "-b", Branch, REPO, SRC], check=True
    )
else:
    subprocess.run(["git", "-C", SRC, "pull", "--ff-only"], check=False)

sys.path.insert(0, f"{SRC}/src")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "pyyaml"], check=False)

from comfycolab import Config, launch

launch(
    Config.from_notebook(
        preset=Preset,
        data_dir=DataDir,
        models=ExtraModels,
        tunnel=Tunnel,
        tunnel_region=Region,
        extra_args=ExtraArgs,
        mount_drive=MountDrive,
        civitai_token=CivitaiToken,
        hf_token=HFToken,
    )
)
```
